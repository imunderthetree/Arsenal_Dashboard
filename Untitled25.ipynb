```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
matches_data = pd.read_csv(f"C:\\Users\\LENOVO\\Downloads\\Visualization Project\\matches.csv")
matches_data.head()
```

```python
matches_data.info()
#data completed
```

```python
matches_data["Season"].unique()
# we have 6 seasons (filter , bar (wins rate) to check wich season had issues , which had greate winning rate and compare )
```

```python
#arsenal score , oponenet , seasons
#collecting all the matches of the same season under the same umbrella
# if arsenal > oponenet
#we wont dismiss the numbers too (maybe arsenal scored high at a certain match but the oponent was too good )
matches_data['comparison']=matches_data['ArsenalScore'].where(
    matches_data['ArsenalScore'] > matches_data['OpponentScore']
)
plt.bar(matches_data['Season'] , matches_data['comparison'])
#so the maximum number of scoring for arsenal for winning was 5
```

```python
#how many times arsenal won in each season
wins_per_season = (
    matches_data
    .groupby('Season')
    .apply(lambda x: (x['ArsenalScore'] > x['OpponentScore']).sum())
)
# used groupby sue to the usage of sum
plt.bar(wins_per_season.index , wins_per_season)
#greatest season was 21/22 and the least sucessful one was 19/20
matches_data['win']=matches_data['ArsenalScore']>matches_data['OpponentScore']
```

```python
#now focus on these two seasons
best=matches_data.where(matches_data['Season']=='2021/22')
least= matches_data.where(matches_data['Season']=='2019/20')
#least.info() # 14 wins
best.info()#22 wins
# tour (which weeks are best and worst / maybe teh opponent)
```

```python
colors = ['green' if w == 1 else 'red' for w in best['win']]

# Set figure size
plt.figure(figsize=(12,1 ))  # width=12 inches, height=6 inches

# Plot
plt.scatter(best['Tour'], best['win'], s=100, c=colors)
plt.xlabel('Tour')
plt.ylabel('Win')
plt.title('Wins per Tour')
plt.show()
```

```python
fig, axes = plt.subplots(6, 1, figsize=(11,6))
for i, season in enumerate(matches_data['Season'].unique()):
    season_data = matches_data[matches_data['Season'] == season]
    colors = ['blue' if w == 1 else 'red' for w in season_data['win']]
    axes[i].scatter(season_data['Tour'], season_data['win'], s=100, c=colors)
    axes[i].set_xlabel('Tour')
    axes[i].set_ylabel(f'{season}' , rotation=0)
fig.lines.append(plt.Line2D([0.5, 0.5], [0, 1], transform=fig.transFigure, color='black', linewidth=4))
plt.tight_layout()
plt.show()
```
